## 0.1 한 장으로 보는 전체 흐름

```text
[1] 원본 데이터          Paderborn 대학 베어링 진동 (.mat, 64 kHz, 4초)
        │  3장
        ▼
[2] 전처리              1 kHz 저역통과 → 4 kHz로 낮춤 → 1초 윈도우로 자르기
        │  4장
        ▼
[3] 데이터 나누기         train / validation / test
        │  5–6장
        ▼
[4] Teacher 모델 학습     MOMENT Base(큰 모델) + LoRA(작은 보정 장치)
        │  7–10장
        ▼
[5] 지식 증류(KD)         Teacher의 판단을 MOMENT Small(작은 모델)이 따라 배움
        │  11–12장
        ▼
[6] ONNX 변환            PyTorch 없이도 돌아가는 범용 모델 파일로 변환
        │  13–14장
        ▼
[7] 라즈베리파이 배포       onnxruntime으로 실시간 추론
        │  15–16장
        ▼
    허깅페이스 공개
```

### 왜 모델을 두 개(Teacher, Student) 만드나요?
- **Teacher (MOMENT Base)**: 크고 똑똑하지만 느립니다. 라즈베리파이에서 돌리기엔 무겁습니다.
- **Student (MOMENT Small)**: 작고 빠릅니다. 대신 혼자 배우면 Teacher만큼 잘하기 어렵습니다.
- **지식 증류(Knowledge Distillation)**: Student가 정답뿐 아니라 Teacher가 "얼마나 확신하는지"까지 보고 배우게 하는 방법입니다(11장).
